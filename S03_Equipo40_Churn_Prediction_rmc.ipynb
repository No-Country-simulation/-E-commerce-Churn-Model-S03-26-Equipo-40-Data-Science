## ANÁLISIS EXPLORATORIO DE DATOS

### Importar Librerías

```python
import os
import kagglehub
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

### Carga de Datos

```python
# DESCARGAR DATASET
url = kagglehub.dataset_download("mashlyn/online-retail-ii-uci")
file_path = os.path.join(url, 'online_retail_II.csv')

# ESTRUCTURA DATASET
# Invoice:      numero de factura
# StockCode:    codigo de producto
# Description:  nombre de producto
# Quantity:     cantidad de producto
# InvoiceDate:  fecha de factura
# Price:        precio unitario de producto
# Customer ID:  codigo de cliente
# Country:      pais del cliente
data_types = {
    'Invoice': 'object',
    'StockCode': 'object',
    'Description': 'object',
    'Quantity': 'int32',
    'Price': 'float64',
    'Customer ID': 'object',
    'Country': 'object'
}
df = pd.read_csv(file_path, dtype=data_types)
```

```python
# INSPECCION INICIAL
df.sample(5)
```

```python
# INSPECCION DEL DATASET
# -Facturas:      53k
# -Cancelaciones:  8k
# -Posiciones de facturas:  1067k
# -Posiciones de cancelac.:   19k
# -Campos: 8
print(df.shape)     # filas y columnas
print(df['Invoice'].nunique())  # facturas
print(df[df['Invoice'].str.contains(r'^C.*', na=False)]['Invoice'].count())    # posiciones de cancelaciones
print(df[df['Invoice'].str.contains(r'^C.*', na=False)]['Invoice'].nunique())  # cancelaciones
```
